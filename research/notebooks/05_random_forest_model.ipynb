```python
SKIP_DIRS = {".git", ".ipynb_checkpoints", "__pycache__", "node_modules",
             ".venv", "venv", "env", ".idea", ".vscode"}

def find_file(filename, required_cols=None, prefer=("raw",), must_exist=True):
    """Cari `filename` di folder kerja, lalu naik ke folder induk sampai root repo (.git).
    Jika required_cols diberikan, hanya file yang header-nya memuat kolom tsb yang dianggap cocok."""
    start = Path.cwd().resolve()
    roots = []
    for p in [start, *start.parents]:
        roots.append(p)
        if (p / ".git").exists() or len(roots) >= 4:
            break

    for root in roots:
        matches = []
        for dirpath, dirnames, filenames in os.walk(root):
            dirnames[:] = [d for d in dirnames if d not in SKIP_DIRS]
            if filename in filenames:
                path = Path(dirpath) / filename
                if required_cols:
                    try:
                        cols = pd.read_csv(path, nrows=0).columns
                    except Exception:
                        continue
                    if not set(required_cols).issubset(cols):
                        continue
                matches.append(path)
        if matches:
            matches.sort(key=lambda p: (not any(k in p.parts for k in prefer), len(p.parts)))
            if len(matches) > 1:
                print(f"Ditemukan {len(matches)} kandidat untuk '{filename}', dipakai yang pertama:")
                for m in matches:
                    print("   -", m)
            return matches[0]

    if must_exist:
        raise FileNotFoundError(
            f"'{filename}' tidak ditemukan. Dicari di: {[str(r) for r in roots]}. "
            "Pastikan notebook dijalankan dari dalam folder repo.")
    return None

DATA_PATH = find_file("train.csv", required_cols=["date", "food_waste_kg"], prefer=("raw",))
print("Dataset mentah dipakai:", DATA_PATH)

# ---------- Konfigurasi ----------
TARGET = "food_waste_kg"
SEED = 42
TEST_FRAC = 0.20        # 20% data terbaru -> test (sama dengan split sebelumnya: 728 train / 183 test)
VAL_FRAC = 0.15         # 15% data terbaru dari train -> validation (untuk tuning)
USE_LAG = False         # lag-1 target hanya bernilai kecil; aktifkan hanya jika ketersediaannya bisa dijustifikasi
CLIP_MEALS = False      # Robust: apakah meals_served ikut di-clip
SEEDS = [0, 1, 2, 3, 4] # seed untuk evaluasi akhir RF

# Folder output ditaruh di root repo (bukan di folder notebooks), dibuat otomatis
OUT_ROOT = Path.cwd().resolve()
for p in [OUT_ROOT, *OUT_ROOT.parents]:
    if (p / ".git").exists():
        OUT_ROOT = p
        break
RESULTS_DIR = OUT_ROOT / "results"
MODELS_DIR = OUT_ROOT / "models"
RESULTS_DIR.mkdir(exist_ok=True)
MODELS_DIR.mkdir(exist_ok=True)
```

```python
df = pd.read_csv(DATA_PATH)
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values(["date", "ID"]).reset_index(drop=True)

print("Shape:", df.shape)
print("Rentang tanggal:", df["date"].min().date(), "->", df["date"].max().date())
print("\nMissing value:\n", df.isna().sum()[df.isna().sum() > 0])
print("\nDuplikat persis (tanpa ID):", df.drop(columns="ID").duplicated().sum())
print("\nNilai unik staff_experience :", df["staff_experience"].unique())
print("Nilai unik waste_category   :", df["waste_category"].unique())

if USE_LAG:
    df["lag1_waste"] = df[TARGET].shift(1)
```

```python
n = len(df)
n_train = int(n * (1 - TEST_FRAC))
train_full = df.iloc[:n_train].copy()
test = df.iloc[n_train:].copy()

n_fit = int(len(train_full) * (1 - VAL_FRAC))
train_fit = train_full.iloc[:n_fit].copy()
val = train_full.iloc[n_fit:].copy()

for nm, d in [("train_fit", train_fit), ("val", val), ("train_full", train_full), ("test", test)]:
    print(f"{nm:10s} n={len(d):4d}  {d['date'].min().date()} -> {d['date'].max().date()}")

assert train_full["date"].max() <= test["date"].min(), "Train tidak boleh lebih baru dari test"
if train_full["date"].max() == test["date"].min():
    print("PERINGATAN: ada tanggal yang sama di batas train/test")
```

```python
NUM_COLS = ["meals_served", "kitchen_staff", "temperature_C", "humidity_percent", "past_waste_kg"]
INT_COLS = ["day_of_week", "special_event"]
CAT_COLS = ["staff_experience", "waste_category"]


class StandardPrep:
    """Standard: konversi tipe, imputasi sederhana (median / modus), one-hot kategori APA ADANYA
    (variasi penulisan tidak dinormalisasi), tanpa clipping, tanpa hapus duplikat.
    Semua statistik dihitung dari data yang di-fit saja (tanpa kebocoran)."""
    name = "standard"

    def __init__(self):
        self.num_cols = NUM_COLS + INT_COLS + (["lag1_waste"] if USE_LAG else [])

    # ---- hook yang dioverride Robust ----
    def _clean_cat(self, d, fit=False):
        for c in CAT_COLS:
            if fit:
                self.modes_[c] = d[c].mode(dropna=True)[0]
            d[c] = d[c].fillna(self.modes_[c])
        return d

    def _fit_extra(self, d): pass
    def _after_numeric(self, d): return d
    def _prepare_train_rows(self, df): return df

    # ---- API ----
    def fit_transform(self, df):
        df = self._prepare_train_rows(df)
        d = df.copy()
        self.modes_ = {}
        for c in self.num_cols:
            d[c] = pd.to_numeric(d[c], errors="coerce")
        self.medians_ = d[self.num_cols].median()
        d[self.num_cols] = d[self.num_cols].fillna(self.medians_)
        self._fit_extra(d)
        d = self._after_numeric(d)
        d = self._clean_cat(d, fit=True)
        self.cat_levels_ = {c: sorted(d[c].dropna().unique()) for c in CAT_COLS}
        return self.transform(df, _fitted_frame=d)

    def transform(self, df, _fitted_frame=None):
        if _fitted_frame is None:
            d = df.copy()
            for c in self.num_cols:
                d[c] = pd.to_numeric(d[c], errors="coerce")
            d[self.num_cols] = d[self.num_cols].fillna(self.medians_)
            d = self._after_numeric(d)
            d = self._clean_cat(d)
        else:
            d = _fitted_frame
        # fitur turunan (sama untuk Standard dan Robust agar perbandingan adil)
        d["is_weekend"] = (d["day_of_week"] >= 5).astype(int)
        d["month"] = pd.to_datetime(d["date"]).dt.month
        d["meals_per_staff"] = d["meals_served"] / d["kitchen_staff"].replace(0, np.nan)
        d["meals_per_staff"] = d["meals_per_staff"].fillna(d["meals_per_staff"].median())
        cols = self.num_cols + ["is_weekend", "month", "meals_per_staff"]
        X = d[cols].copy()
        for c, levels in self.cat_levels_.items():
            for lv in levels:
                X[f"{c}={lv}"] = (d[c] == lv).astype(int)   # kategori tak dikenal -> semua 0
        y = d[TARGET]
        return X, y


class RobustPrep(StandardPrep):
    """Robust: normalisasi kategori (strip + lowercase), missing kategori -> 'unknown',
    hapus duplikat persis HANYA pada data training, winsorizing outlier fitur (batas dari train,
    diterapkan KONSISTEN ke train & test). Target tidak pernah diubah."""
    name = "robust"
    IQR_K = 3.0

    def _clean_cat(self, d, fit=False):
        for c in CAT_COLS:
            d[c] = d[c].astype("string").str.strip().str.lower().fillna("unknown").astype(str)
        return d

    def _prepare_train_rows(self, df):
        key = [c for c in df.columns if c != "ID"]
        out = df.drop_duplicates(subset=key)
        self.n_dup_removed_ = len(df) - len(out)
        return out

    def _fit_extra(self, d):
        clip_cols = ["kitchen_staff", "temperature_C", "humidity_percent", "past_waste_kg"]
        if CLIP_MEALS:
            clip_cols.append("meals_served")
        self.clip_bounds_ = {}
        for c in clip_cols:
            q1, q3 = d[c].quantile(0.25), d[c].quantile(0.75)
            iqr = q3 - q1
            self.clip_bounds_[c] = (q1 - self.IQR_K * iqr, q3 + self.IQR_K * iqr)

    def _after_numeric(self, d):
        for c, (lo, hi) in self.clip_bounds_.items():
            d[c] = d[c].clip(lo, hi)
        return d
```

```python
PREPS = {"standard": StandardPrep, "robust": RobustPrep}
TUNE, FINAL, PREP_FINAL = {}, {}, {}

for name, P in PREPS.items():
    # (a) untuk tuning: fit di train_fit, evaluasi di val
    p_tune = P()
    Xtr, ytr = p_tune.fit_transform(train_fit)
    Xva, yva = p_tune.transform(val)
    TUNE[name] = (Xtr, ytr, Xva, yva)
    # (b) untuk evaluasi akhir: fit di train_full, evaluasi di test (dipakai sekali)
    p_fin = P()
    Xtf, ytf = p_fin.fit_transform(train_full)
    Xte, yte = p_fin.transform(test)
    FINAL[name] = (Xtf, ytf, Xte, yte)
    PREP_FINAL[name] = p_fin
```

```python
rows = []
for name in PREPS:
    Xtf, ytf, Xte, yte = FINAL[name]
    Xtr, ytr, Xva, yva = TUNE[name]
    assert list(Xtf.columns) == list(Xte.columns) and list(Xtr.columns) == list(Xva.columns)
    assert Xtf.isna().sum().sum() == 0 and Xte.isna().sum().sum() == 0, "Masih ada NaN"
    assert "ID" not in Xtf.columns and "date" not in Xtf.columns
    # cek bug lama: special_event harus punya nilai 0 DAN 1 di train maupun test
    assert Xtf["special_event"].nunique() == 2 and Xte["special_event"].nunique() == 2, "special_event hilang!"
    rows.append({
        "preprocessing": name, "n_train": len(Xtf), "n_test": len(Xte), "n_fitur": Xtf.shape[1],
        "event_train": int(Xtf["special_event"].sum()), "event_test": int(Xte["special_event"].sum()),
        "dup_dihapus": getattr(PREP_FINAL[name], "n_dup_removed_", 0),
        "meals_max_train": Xtf["meals_served"].max(), "meals_max_test": Xte["meals_served"].max(),
    })
print(pd.DataFrame(rows).to_string(index=False))

# test set harus IDENTIK antara Standard dan Robust (perbandingan adil)
assert FINAL["standard"][3].index.equals(FINAL["robust"][3].index)
assert np.allclose(FINAL["standard"][3].values, FINAL["robust"][3].values)
print("\nOK: baris & target test identik di kedua preprocessing")

print("\nBatas clipping Robust (dari train):")
for c, (lo, hi) in PREP_FINAL["robust"].clip_bounds_.items():
    n_tr = ((train_full[c] < lo) | (train_full[c] > hi)).sum()
    n_te = ((test[c] < lo) | (test[c] > hi)).sum()
    print(f"  {c:18s} [{lo:8.2f}, {hi:8.2f}]  di luar batas: train={n_tr}, test={n_te}")

print("\nKolom Standard:", list(FINAL["standard"][0].columns))
print("Kolom Robust  :", list(FINAL["robust"][0].columns))
```

```python
def metrics(y_true, y_pred):
    return {"MAE": mean_absolute_error(y_true, y_pred),
            "RMSE": mean_squared_error(y_true, y_pred) ** 0.5,
            "R2": r2_score(y_true, y_pred)}

grid = list(ParameterGrid({
    "n_estimators": [300, 500],
    "max_depth": [None, 10, 20],
    "min_samples_leaf": [1, 2, 5],
    "max_features": ["sqrt", 0.5, 1.0],
}))

best_params, tune_rows = {}, []
for name, (Xtr, ytr, Xva, yva) in TUNE.items():
    best = (np.inf, None)
    for params in grid:
        m = RandomForestRegressor(**params, random_state=SEED, n_jobs=-1).fit(Xtr, ytr)
        s = metrics(yva, m.predict(Xva))
        tune_rows.append({"preprocessing": name, "params": str(params), **{f"val_{k}": v for k, v in s.items()}})
        if s["MAE"] < best[0]:
            best = (s["MAE"], params)
    best_params[name] = best[1]
    print(f"[{name}] best val MAE = {best[0]:.3f}  params = {best[1]}")

tune_df = pd.DataFrame(tune_rows)
tune_df.to_csv(RESULTS_DIR / "rf_tuning_validation.csv", index=False)
tune_df.sort_values("val_MAE").groupby("preprocessing").head(5)
```

```python
final_rows, final_models = [], {}

for name, (Xtf, ytf, Xte, yte) in FINAL.items():
    per_seed = []
    for s in SEEDS:
        m = RandomForestRegressor(**best_params[name], random_state=s, n_jobs=-1)
        t0 = time.perf_counter(); m.fit(Xtf, ytf); t_train = time.perf_counter() - t0
        t0 = time.perf_counter(); pred = m.predict(Xte); t_pred = time.perf_counter() - t0
        per_seed.append({**metrics(yte, pred), "train_time_s": t_train, "pred_time_s": t_pred})
        if s == SEEDS[0]:
            final_models[name] = (m, pred)
    ps = pd.DataFrame(per_seed)
    final_rows.append({"model": "RandomForest", "preprocessing": name,
                       **{f"{c}_mean": ps[c].mean() for c in ps.columns},
                       **{f"{c}_std": ps[c].std() for c in ["MAE", "RMSE", "R2"]}})

# pembanding: tebak rata-rata & regresi linear (Ridge)
for name, (Xtf, ytf, Xte, yte) in FINAL.items():
    for label, mdl in [("Baseline-Mean", DummyRegressor(strategy="mean")),
                       ("Ridge", make_pipeline(StandardScaler(), Ridge(alpha=1.0)))]:
        t0 = time.perf_counter(); mdl.fit(Xtf, ytf); t_train = time.perf_counter() - t0
        t0 = time.perf_counter(); pred = mdl.predict(Xte); t_pred = time.perf_counter() - t0
        final_rows.append({"model": label, "preprocessing": name,
                           **{f"{k}_mean": v for k, v in metrics(yte, pred).items()},
                           "train_time_s_mean": t_train, "pred_time_s_mean": t_pred})

results = pd.DataFrame(final_rows)
results.to_csv(RESULTS_DIR / "rf_experiment_results.csv", index=False)
print(results.round(3).to_string(index=False))

for name, (m, _) in final_models.items():
    joblib.dump({"prep": PREP_FINAL[name], "model": m, "params": best_params[name]},
                MODELS_DIR / f"rf_{name}.joblib")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharey=True)
for ax, (name, (m, _)) in zip(axes, final_models.items()):
    imp = pd.Series(m.feature_importances_, index=FINAL[name][0].columns).sort_values().tail(10)
    imp.plot.barh(ax=ax, title=f"Impurity importance - RF ({name})")
plt.tight_layout(); plt.savefig(RESULTS_DIR / "rf_feature_importance.png", dpi=120); plt.show()

for name, (m, _) in final_models.items():
    Xte, yte = FINAL[name][2], FINAL[name][3]
    pi = permutation_importance(m, Xte, yte, scoring="neg_mean_absolute_error",
                                n_repeats=10, random_state=SEED, n_jobs=-1)
    top = pd.Series(pi.importances_mean, index=Xte.columns).sort_values(ascending=False).head(8)
    print(f"\nPermutation importance (kenaikan MAE jika fitur diacak) - {name}")
    print(top.round(3).to_string())
```

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 8))
for j, (name, (m, pred)) in enumerate(final_models.items()):
    yte = FINAL[name][3]
    axes[0, j].scatter(yte, pred, s=14, alpha=0.6)
    lim = [yte.min(), yte.max()]
    axes[0, j].plot(lim, lim, "r--"); axes[0, j].set_title(f"Aktual vs Prediksi ({name})")
    axes[0, j].set_xlabel("Aktual (kg)"); axes[0, j].set_ylabel("Prediksi (kg)")
    axes[1, j].scatter(pred, yte - pred, s=14, alpha=0.6); axes[1, j].axhline(0, color="r", ls="--")
    axes[1, j].set_title(f"Residual ({name})")
    axes[1, j].set_xlabel("Prediksi (kg)"); axes[1, j].set_ylabel("Residual (kg)")
plt.tight_layout(); plt.savefig(RESULTS_DIR / "rf_actual_vs_pred.png", dpi=120); plt.show()

# error pada baris dengan meals_served ekstrem (>1000) vs normal
for name, (m, pred) in final_models.items():
    Xte, yte = FINAL[name][2], FINAL[name][3]
    ext = Xte["meals_served"] > 1000
    print(f"[{name}] MAE meals<=1000: {mean_absolute_error(yte[~ext], pred[~ext.values]):.2f} (n={int((~ext).sum())})"
          f" | meals>1000: {mean_absolute_error(yte[ext], pred[ext.values]):.2f} (n={int(ext.sum())})")
```

## Langkah 4 Menyiapkan fitur dan baris evaluasi yang setara
Statistik preprocessing dipelajari dari training saja, lalu dipakai pada validasi/test. Tujuh observasi pertama diperlukan untuk jendela LSTM. Enam baris awal tidak dijadikan target pada seluruh model agar baris pembanding identik. Kategori CatBoost tetap berupa teks; RF/XGBoost memakai one-hot. LSTM memakai urutan fitur terskala. Tidak ada label test yang digunakan untuk konteks.

```python
eligible = np.arange(len(df)) >= WINDOW-1
idx = {s: np.flatnonzero((df.split==s).to_numpy() & eligible) for s in ['train','validation','test']}
data = {}
for method in ['standard','robust']:
    prep = FoodPreprocessor(method).fit(df[df.split=='train'])
    x = prep.transform(df)
    scaled = prep.transform(df, scaled=True)
    native = prep.native(df)
    target_scaler = StandardScaler().fit(df.loc[idx['train'],TARGET].to_numpy().reshape(-1,1))
    y = {s:df.loc[ix,TARGET].to_numpy() for s,ix in idx.items()}
    if MODEL=='LSTM':
        features = {s:sequence(scaled,ix) for s,ix in idx.items()}
    elif MODEL=='CatBoost':
        features = {s:native.iloc[ix] for s,ix in idx.items()}
    else:
        features = {s:x.iloc[ix] for s,ix in idx.items()}
    assert np.isfinite(x.to_numpy()).all() and np.isfinite(scaled).all()
    assert not set(['ID',TARGET,'waste_category','past_waste_kg']).intersection(x.columns)
    data[method] = {'prep':prep,'ys':target_scaler,'y':y,'features':features,'names':native.columns.tolist() if MODEL=='CatBoost' else x.columns.tolist()}
    print(method, '| jumlah fitur:', x.shape[1], '| bentuk input:', {s:v.shape for s,v in features.items()})
    display(native.head(4))
```

## Langkah 5 Menulis cara pelatihan model
Fungsi berikut berisi kode model yang benar-benar dilatih, bukan placeholder. Training memperbarui parameter dari data latihan. Validasi memantau kemampuan di periode berikutnya. Early stopping menghentikan jaringan/boosting ketika validasi tidak membaik, lalu memakai keadaan terbaik. RF melihat kestabilan error ketika jumlah pohon bertambah.

Dua konfigurasi diuji per metode, bukan pencarian optimum global. Learning rate adalah besar langkah belajar; kedalaman pohon mengatur kerumitan aturan; regularisasi/minimum daun membatasi model agar tidak terlalu menyesuaikan training.

```python
def latih_kandidat(config, xtr, ytr, xva, yva, target_scaler):
    start = time.perf_counter()
    model = RandomForestRegressor(random_state=SEED, n_jobs=2, warm_start=True, **config)
    rows = []
    for n in [20, 60, 120, 200]:
        model.set_params(n_estimators=n).fit(xtr, ytr)
        rows.append({'step': n, 'train_RMSE': metrics(ytr, model.predict(xtr))['RMSE'], 'validation_RMSE': metrics(yva, model.predict(xva))['RMSE']})
    curve = pd.DataFrame(rows)
    return model, curve, time.perf_counter() - start
```

## Langkah 6 Mencoba kandidat dan menghitung skor validasi
Bagian ini membutuhkan waktu komputasi. Semua kandidat dilatih hanya pada training. Masing-masing dibandingkan menggunakan MAE validasi dalam kg. Data test belum digunakan untuk memilih konfigurasi. Kurva training/validasi dan waktu setiap kandidat dicatat agar prosesnya dapat diaudit.

```python
candidates = {}
tuning_rows = []
for method, d in data.items():
    candidates[method] = []
    for config in CONFIGS:
        model, curve, duration = latih_kandidat(config, d['features']['train'], d['y']['train'], d['features']['validation'], d['y']['validation'], d['ys'])
        prediction = predict(model, d['features']['validation'], MODEL, d['ys'])
        score = metrics(d['y']['validation'], prediction)
        tuning_rows.append({'preprocessing':method, 'config':json.dumps(config), **score, 'train_time_s':duration})
        candidates[method].append({'model':model,'curve':curve,'duration':duration,'config':config,'score':score})
        print(method, config, '| MAE validasi:', round(score['MAE'],4), 'kg', flush=True)
tuning = pd.DataFrame(tuning_rows)
display(tuning.sort_values(['preprocessing','MAE']))
```

## Langkah 7 Mengunci konfigurasi dan membaca kurva belajar
MAE validasi terendah memilih satu kandidat pada Standard dan satu pada Robust. Kurva yang turun berarti error mengecil; kurva tidak harus selalu turun. Error training yang jauh lebih rendah daripada validasi bisa menandakan overfitting atau perubahan distribusi waktu. Sumbu RF adalah jumlah pohon, sumbu boosting adalah tahap, dan sumbu LSTM adalah epoch. RF tidak memakai gradient descent.

```python
selected = {method:min(items,key=lambda item:item['score']['MAE']) for method,items in candidates.items()}
for method,item in selected.items():
    label=f'{MODEL}_{method}'
    item['curve'].to_csv(OUT/f'{label}_curve.csv',index=False)
    tuning[tuning.preprocessing==method].drop(columns='preprocessing').to_csv(OUT/f'{label}_tuning.csv',index=False)
    plot_curve(item['curve'],label)
    print('Terpilih:',method,item['config'])
    display(Image(filename=str(OUT/f'{label}_curve.png')))
```

## Langkah 8 Mengukur seluruh metrik training validasi dan test
MAE: kesalahan absolut rata-rata dalam kg. RMSE: lebih sensitif terhadap error besar. R²: kemampuan menjelaskan variasi; dapat negatif. MedAE: median error absolut. Bias: prediksi dikurangi aktual; positif berarti terlalu tinggi. WAPE: total error absolut dibanding total limbah, bukan akurasi klasifikasi.

Waktu prediksi berikut diukur untuk satu batch tanpa preprocessing dan pemuatan model. Milidetik per baris adalah waktu batch dibagi jumlah baris, bukan benchmark permintaan tunggal. Waktu training mencakup pelatihan dan pemantauan kurva kandidat terpilih, termasuk impor awal library/pengukuran gradien; total tuning mencakup kedua kandidat. Model tidak dilatih ulang pada validasi.

```python
result_rows = []
protocol = hashlib.sha256((ROOT/'src/experiment.py').read_bytes()+(ROOT/'dataset/raw/train.csv').read_bytes()).hexdigest()
for method,item in selected.items():
    d=data[method];label=f'{MODEL}_{method}'
    row={'model':MODEL,'preprocessing':method,'config':json.dumps(item['config']),'protocol_sha256':protocol,
         'train_time_s':item['duration'],'tuning_total_time_s':float(tuning.loc[tuning.preprocessing==method,'train_time_s'].sum()),
         **{f'n_{s}':len(v) for s,v in d['y'].items()}}
    for split in ['train','validation','test']:
        start=time.perf_counter()
        prediction=predict(item['model'],d['features'][split],MODEL,d['ys'])
        elapsed=time.perf_counter()-start
        row.update({f'{split}_{k}':v for k,v in metrics(d['y'][split],prediction).items()})
        row[f'{split}_prediction_time_s']=elapsed
        row[f'{split}_ms_per_row']=elapsed/len(prediction)*1000
        pd.DataFrame({'row':df.loc[idx[split],'_row'],'date':df.loc[idx[split],'date'],'actual':d['y'][split],
                      'prediction':prediction,'residual':d['y'][split]-prediction}).to_csv(OUT/f'{label}_{split}_predictions.csv',index=False)
    save_json(OUT/f'{label}_metrics.json',row)
    result_rows.append(row)
hasil=pd.DataFrame(result_rows)
for split in ['train','validation','test']:
    display(Markdown(f'### Metrik {split}'))
    display(hasil[['model','preprocessing']+[f'{split}_{m}' for m in ['MAE','RMSE','R2','MedAE','Bias','WAPE_percent']]])
display(hasil[['model','preprocessing','train_time_s','tuning_total_time_s','test_prediction_time_s','test_ms_per_row']])
```

## Langkah 10 Menilai fitur yang membantu prediksi
Permutation importance mengacak satu fitur pada validasi sebanyak tiga kali. Kenaikan MAE positif berarti fitur tersebut membantu model. Nilai negatif dapat terjadi akibat noise atau model yang terlalu mengandalkan fitur tersebut. Ini bukan bukti kausal. Rasio dan porsi saling berkaitan sehingga importance dapat terbagi. Pada LSTM, satu jalur fitur sepanjang jendela diacak sebagai satu blok; hasil bersifat diagnostik.

```python
for method,item in selected.items():
    d=data[method];rng=np.random.default_rng(SEED);importance=[]
    for j,feature in enumerate(d['names']):
        differences=[]
        for repeat in range(3):
            shuffled=d['features']['validation'].copy();order=rng.permutation(len(shuffled))
            if MODEL=='LSTM':shuffled[:,:,j]=shuffled[order,:,j]
            else:shuffled.iloc[:,j]=shuffled.iloc[order,j].to_numpy()
            score=metrics(d['y']['validation'],predict(item['model'],shuffled,MODEL,d['ys']))['MAE']
            differences.append(score-item['score']['MAE'])
        importance.append({'feature':feature,'MAE_increase':np.mean(differences),'std':np.std(differences)})
    imp=pd.DataFrame(importance).sort_values('MAE_increase',ascending=False)
    imp.to_csv(OUT/f'{MODEL}_{method}_importance.csv',index=False)
    display(imp)
    top=imp.head(12).sort_values('MAE_increase')
    fig,ax=plt.subplots(figsize=(9,4));ax.barh(top.feature,top.MAE_increase,xerr=top['std'])
    ax.set(xlabel='Kenaikan MAE validasi kg',title=f'Pengaruh fitur {MODEL} {method}')
    fig.tight_layout();display(fig);plt.close(fig)
```

## Langkah 13 Membuka kembali artefak untuk melanjutkan pekerjaan
Contoh ini memeriksa model tersimpan pada lima observasi test yang sama, bukan pengujian baru. Jika prediksinya sama, pasangan model/preprocessing dapat dibuka kembali tanpa training. Saat dipakai pada input baru, urutan kolom dan asumsi ketersediaan fitur harus tetap sama. Model LSTM membutuhkan konteks, bukan hanya satu baris.

```python
for method in ['standard','robust']:
    label=f'{MODEL}_{method}'
    bundle=joblib.load(ART/f'{label}_preprocessing.joblib')
    if MODEL=='LSTM':
        import tensorflow as tf
        loaded=tf.keras.models.load_model(ART/f'{label}.keras',compile=False)
        encoded=bundle['preprocessor'].transform(df,scaled=True)
        sample=sequence(encoded,idx['test'][:5])
    else:
        loaded=joblib.load(ART/f'{label}.joblib')
        transformed=bundle['preprocessor'].native(df) if MODEL=='CatBoost' else bundle['preprocessor'].transform(df)
        sample=transformed.iloc[idx['test'][:5]]
    restored=predict(loaded,sample,MODEL,bundle['target_scaler'])
    saved=pd.read_csv(OUT/f'{label}_test_predictions.csv').prediction.iloc[:5].to_numpy()
    np.testing.assert_allclose(restored,saved,rtol=1e-5,atol=1e-5)
    display(pd.DataFrame({'prediksi_tersimpan':saved,'prediksi_setelah_dibuka':restored}))
    print(method,': LULUS, model dapat dibuka kembali')
```
